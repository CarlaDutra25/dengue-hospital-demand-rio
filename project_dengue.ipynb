# **Resumo**

Este projeto propõe um produto analítico preditivo para apoiar a gestão hospitalar frente a surtos de arboviroses (dengue, zika e chikungunya) no município do Rio de Janeiro. Aplicando técnicas de séries temporais para prever a evolução semanal de casos e internações a partir de dados epidemiológicos e climáticos, traduzindo essas previsões em um indicador operacional de nível de alerta hospitalar (Normal, Atenção, Crítico), calibrado com a capacidade instalada de leitos sendo validado por profissionais da Secretaria de Saúde do Rio de Janeiro, garantindo o rigor e o impacto social da solução proposta.


# **Introdução**

A aplicação de técnicas de análise e modelagem de séries temporais na saúde pública constitui uma ferramenta estratégica na compreensão e no combate a doenças infectocontagiosas. No município do Rio de Janeiro, o caráter epidêmico recorrente de arboviroses como dengue, zika e chikungunya acarreta forte pressão sobre a rede hospitalar pública e privada: os surtos provocam picos repentinos de atendimento e internação, sobrecarregando a infraestrutura de saúde e dificultando o planejamento antecipado de leitos, escalas médicas e aquisição de insumos críticos.

A proliferação do vetor *Aedes aegypti* e a dinâmica de contágio no território fluminense mantêm relação direta com variáveis meteorológicas (temperatura e precipitação acumulada) e com padrões sazonais específicos da região. A construção de modelos preditivos que relacionem condicionantes climáticos defasados à evolução de casos e internações viabiliza a transição de uma gestão hospitalar reativa para uma atuação preventiva e otimizada, que seria a motivação central deste projeto.

**Objetivo geral:** desenvolver e validar um produto analítico preditivo em duas camadas:

1. Modelagem temporal multivariada para estimar a incidência de casos e internações por arboviroses no Rio de Janeiro com base em fatores climáticos e sazonalidade
2. Um modelo de classificação operacional que categoriza o nível de alerta de demanda hospitalar (Normal, Atenção, Crítico), cruzando as previsões temporais com a capacidade instalada de leitos do SUS.

**Objetivos específicos:**

1. Consolidar e harmonizar os registros históricos epidemiológicos e meteorológicos do Rio de Janeiro na escala de Semana Epidemiológica (2015–2025)
2. Caracterizar os componentes estruturais das séries temporais (tendência, sazonalidade, ciclos e autocorrelação)
3. Implementar e comparar algoritmos preditivos estatísticos e de aprendizado profundo (SARIMAX e LSTM/RNN)
4. Construir e calibrar o modelo de classificação de risco hospitalar com o suporte técnico de especialistas do SUS-RJ
5. Disponibilizar os resultados em uma interface de documentação analítica interativa no GitHub

**Base de dados:** composta por séries temporais multivariadas organizadas segundo o calendário oficial de Semanas Epidemiológicas, cobrindo o período contínuo de 2015 a 2025 no Rio de Janeiro. As fontes utilizadas são:

- **InfoDengue (Fiocruz/FGV):** histórico semanal de notificações estimadas e confirmadas de dengue, zika e chikungunya, taxa de incidência e níveis regionais de alerta
- **DATASUS (SINAN/SIH-SUS):** notificações de agravos e Autorizações de Internação Hospitalar (AIH) decorrentes de arboviroses na rede pública
- **INMET/Open-Meteo:** temperatura média, temperatura máxima e precipitação pluviométrica acumulada semanais
- **CNES:** capacidade instalada de leitos de enfermaria e UTI, utilizada como referência para a classificação de níveis de alerta hospitalar

**Justificativa:** embora existam plataformas consolidadas de monitoramento epidemiológico no Brasil - como o próprio InfoDengue - este projeto se diferencia em três pontos:

1. **Foco na demanda hospitalar operacional**, e não apenas no risco de notificação/transmissão populacional, ao modelar a conversão de contágios em internações e necessidade real de leitos
2. **Arquitetura em duas camadas**, integrando a previsão quantitativa de séries temporais (SARIMAX e LSTM) a uma camada de classificação funcional de alerta hospitalar
3. **Validação prática com especialistas da rede do SUS-RJ**, com ciclos formais de calibração técnica ao longo do desenvolvimento.

O projeto alinha-se às diretrizes da Agenda 2030 da ONU, contemplando diretamente:

- **ODS 3 - Saúde e Bem Estar:** Ao fornecer instrumentos de inteligência preditiva para apoiar a gestão hospitalar e a resposta rápida a epidemias.
- **ODS 11 - Cidades e Comunidades Sustentáveis:** Ao fortalecer a resiliência dos sistemas urbanos de saúde frente a variabilidades climáticas e surtos sazonais em grandes metrópoles.


# **Referencial Teórico**

A modelagem preditiva de doenças de transmissão vetorial a partir de séries temporais é um campo consolidado, com abordagens que evoluíram de modelos estatísticos clássicos para técnicas híbridas que combinam estatística e aprendizado profundo.

**1. Modelos estatísticos clássicos (SARIMA/SARIMAX)**

- O modelo **SARIMA/SARIMAX** (*Seasonal Autoregressive Integrated Moving Average with eXogenous variables*), fundamentado em Box e Jenkins (1970), continua sendo referência para previsão epidemiológica.
- Martinez e Silva (2011), em estudo com dados de dengue em Campinas (SP), validaram sua aplicabilidade a séries de casos com forte componente sazonal.
- **Vantagem:** interpretabilidade dos coeficientes e baixo custo computacional.
- **Limitação:** suposição de relações predominantemente lineares entre a série e as variáveis exógenas, o que pode não capturar toda a complexidade de fenômenos climático-epidemiológicos.

**2. Modelos de aprendizado profundo (LSTM)**

- Redes **LSTM** (*Long Short-Term Memory*), propostas por Hochreiter e Schmidhuber (1997), vêm sendo aplicadas ao mesmo problema por sua capacidade de capturar dependências não lineares e de longo prazo.
- Sahd et al. (2026) compararam diretamente SARIMAX e LSTM para previsão de dengue em Londrina (PR), incorporando temperatura e umidade como variáveis exógenas.
- **Resultado:** o SARIMAX apresentou desempenho mais estável, enquanto a LSTM, apesar de maior volatilidade nas previsões, mostrou-se promissora, o que motiva a comparação entre as duas técnicas também neste projeto, em vez de se adotar uma única abordagem a priori.

**3. Trabalhos aplicados ao Rio de Janeiro e à demanda hospitalar**

- A literatura recente já avaliou modelos estatísticos e de aprendizado de máquina para previsão de dengue especificamente no Rio de Janeiro, reforçando a viabilidade metodológica do recorte geográfico escolhido.
- Também há trabalhos que investigam especificamente a previsão de **internações** por dengue no Brasil a partir de dados climáticos, considerando os atrasos reais de notificação (*reporting delays*) — desafio prático também enfrentado neste projeto, dado o uso de dados de AIH/SIH-SUS.
- *(As referências completas desses dois trabalhos estão em processo de validação de autoria e serão incorporadas na próxima atualização da bibliografia.)*

**4. Técnicas de decomposição e transformação**

- A **decomposição STL** (*Seasonal-Trend decomposition using Loess*), proposta por Cleveland et al. (1990), foi adotada para caracterizar a série por ser mais robusta a sazonalidade de amplitude variável o que é relevante neste caso, já que surtos de arbovirose não apresentam a mesma intensidade sazonal ano a ano.
- A estabilização de variância da série de casos foi realizada pela **transformação de Box-Cox** (Box e Cox, 1964), com o parâmetro ótimo estimado pelo método de Guerrero (1993).

**5. Diferencial do projeto**

A maior parte da literatura correlata concentra-se em prever **incidência de casos**, não a **demanda operacional de leitos**. Este projeto se diferencia ao:

- Integrar a previsão de séries temporais (SARIMAX/LSTM) a uma camada de classificação de risco hospitalar
- Calibrar essa classificação com a capacidade instalada de leitos (CNES)
- Validar o resultado com profissionais do SUS-RJ, traduzindo a previsão epidemiológica em uma decisão operacional direta para a gestão hospitalar

# **Diagrama de Solução**

A solução é estruturada em cinco etapas sequenciais:

1. **Coleta de dados**: 
- InfoDengue (Fiocruz/FGV, casos semanais de arbovirose)
- Open-Meteo/INMET (temperatura e precipitação)
- DATASUS SINAN/SIH-SUS (notificações e internações hospitalares)
- CNES (capacidade instalada de leitos), todos referentes ao município do Rio de Janeiro, período 2015–2025

2. **Pré-processamento**:
- Reamostragem para a escala de Semana Epidemiológica
- Imputação de eventuais lacunas por interpolação temporal
- Estabilização de variância por transformação de Box-Cox

3. **Análise exploratória**:
- Verificação de estacionariedade (teste ADF)
- Funções de autocorrelação (ACF/PACF)
- Decomposição da série (clássica e STL) para caracterizar tendência, sazonalidade e ruído
- Investigação da relação defasada entre variáveis climáticas e casos

4. **Modelagem preditiva**: 
- Implementação e comparação de SARIMAX (com variáveis exógenas climáticas defasadas) e LSTM, avaliados por métricas de erro (ex.: MAE, RMSE) sobre um conjunto de teste out-of-sample.

5. **Classificação e validação**:
- Cruzamento das previsões de casos/internações com a capacidade instalada de leitos (CNES) para gerar um indicador de nível de alerta hospitalar (Normal / Atenção / Crítico), com validação e calibração junto a profissionais do SUS-RJ
- Disponibilização dos resultados em documentação analítica interativa no GitHub.

# **EDA e Pré-processamento dos dados**

A análise exploratória completa está no notebook [Etapa 2 — Análise Exploratória](https://github.com/CarlaDutra25/PROJETO-APLICADO-IV/blob/main/Etapa2_Analise_Exploratoria2_com_correlacao_climatica.ipynb), e a preparação final dos dados no notebook [Etapa 3 — Modelagem](https://github.com/CarlaDutra25/PROJETO-APLICADO-IV/blob/main/Etapa3_Modelagem_SARIMAX.ipynb). Os principais pontos são:

**Dados e recortes**
- Séries semanais de 2015 a 2025 do município do Rio de Janeiro: casos estimados de dengue, chikungunya e zika (InfoDengue) e temperatura e precipitação diárias (Open-Meteo), agregadas por semana epidemiológica.
- No período em que as três séries existem (2015–fev/2024), a dengue representa cerca de 60% dos casos, a chikungunya 28% e a zika 12%.
- A API do InfoDengue não retorna dados de zika para o município após fevereiro de 2024; por isso, a variável-alvo é a soma de dengue e chikungunya, e a zika aparece apenas na análise descritiva.

**Qualidade e preparação**
- Correção do alinhamento das semanas: o clima passou a ser agregado de domingo a sábado e rotulado pelo domingo de início da semana epidemiológica, como no InfoDengue (antes havia uma defasagem oculta de cerca de uma semana).
- Lacunas internas imputadas por interpolação temporal; nenhum dado sintético é usado nos resultados.
- Transformação de Box-Cox com λ ≈ −0,02, praticamente logarítmica; por isso foi adotado `y = log(1 + casos)`.

**Características da série**
- Teste ADF: série estacionária (p < 0,05).
- Decomposição STL com período de 52 semanas: tendência forte (F_T ≈ 0,75) e sazonalidade anual moderada (F_S ≈ 0,41), o que justifica o termo sazonal S = 52.
- Correlação cruzada: a temperatura apresenta a maior correlação com os casos com cerca de 11 semanas de defasagem (r ≈ 0,45); a precipitação tem correlação fraca (r ≈ 0,16).

**Dados hospitalares (Etapa 4)**
- Internações e dias de permanência do SIH-SUS por residência no município, leitos SUS do CNES por especialidade e casos notificados de dengue por Área de Planejamento (Tabnet Municipal SMS-Rio), além da população de 2022 por bairro (Data.Rio), mapeada para as 10 Áreas de Planejamento.

# **Modelos**

**Previsão de casos** ([notebook da Etapa 3](https://github.com/CarlaDutra25/PROJETO-APLICADO-IV/blob/main/Etapa3_Modelagem_SARIMAX.ipynb))
- Modelos comparados: ARIMAX com termos de Fourier (com e sem clima), SARIMAX(1,0,1)(1,0,0)₅₂ com clima defasado e LSTM (opcional).
- Baselines: persistência (repetir o último valor observado) e ingênuo sazonal (repetir a mesma semana do ano anterior).
- Validação *walk-forward* com origem móvel em três anos de teste (2023, 2024 e 2025): para cada ano, defasagens climáticas, padronização, ordem do modelo e parâmetros são definidos apenas com os anos anteriores, sem vazamento de informação. As previsões são feitas de 1 a 4 semanas à frente, com intervalos de previsão de 80%.
- O clima entra apenas com defasagem maior ou igual ao horizonte de previsão, para que esteja disponível no momento da previsão.

**Demanda hospitalar, alerta e Áreas de Planejamento** ([notebook da Etapa 4](https://github.com/CarlaDutra25/PROJETO-APLICADO-IV/blob/main/Etapa4_Demanda_Alerta_AP.ipynb))
- Casos previstos são convertidos em internações e leitos ocupados com a taxa de internação por caso e a permanência média do SIH-SUS nos 12 meses anteriores à previsão (Lei de Little: leitos ≈ internações por semana × permanência ÷ 7).
- A ocupação prevista é comparada com os leitos SUS de clínica geral e pediatria clínica do CNES para gerar o alerta Normal / Atenção / Crítico (limiares provisórios de 1% e 2% dos leitos, a calibrar com a SMS-RJ).
- A previsão municipal é desagregada para as 10 Áreas de Planejamento pela participação de cada área nos casos notificados (*top-down*), e avalia-se a capacidade de identificar com antecedência as áreas de maior incidência.

# **Dicussão e Conclusão**

**Principais achados**
- Modelos estatísticos clássicos acompanham bem a série de casos, mas antecipam pouco: o ganho sobre a persistência só aparece com 3 a 4 semanas de antecedência e é modesto (≈ 9%).
- O valor operacional está na tradução da previsão em decisão: intervalos de previsão confiáveis, níveis de alerta e estimativa de leitos validada com dados hospitalares reais.
- Mesmo no pico da epidemia de 2024, a ocupação média ficou em torno de 3% dos leitos clínicos e pediátricos SUS da cidade. A pressão hospitalar é concentrada em territórios e semanas específicas, e não uma falta de capacidade do município como um todo; isso reforça a importância da visão por Área de Planejamento.

**Limitações**
- Alerta avaliado em poucos meses de pressão real (fevereiro a maio de 2024).
- Taxa de internação e permanência municipais, aplicadas igualmente a todas as Áreas de Planejamento; leitos por área e ocupação real não estão disponíveis em dados públicos tabulados.
- Participação por área baseada em casos notificados de dengue a partir de 2023; zika fora do alvo; limiares de alerta provisórios.
- O grupo "restante de outras febres por arbovírus" do SIH inclui outras arboviroses além da chikungunya.

**Próximos passos**
- Calibração dos limiares e validação com a equipe técnica da Secretaria Municipal de Saúde do Rio de Janeiro.
